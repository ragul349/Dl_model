```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

print("TensorFlow version:", tf.__version__)
print("Libraries imported successfully!")
```

```python
import os
import zipfile
import pandas as pd

# Check if a zip file was uploaded and extract it
zip_files = [f for f in uploaded.keys() if f.lower().endswith('.zip')]
if zip_files:
    print(f"Extracting {zip_files[0]}...")
    with zipfile.ZipFile(zip_files[0], 'r') as zip_ref:
        zip_ref.extractall('.')

# Now search for CSV files in the current directory or uploaded files
csv_files = [f for f in os.listdir('.') if f.lower().endswith('.csv')]
if not csv_files:
    csv_files = [f for f in uploaded.keys() if f.lower().endswith('.csv')]

if not csv_files:
    raise ValueError("No CSV file found in the uploaded files or the extracted archive.")

target_csv = csv_files[0]
print(f"Loading {target_csv}...")
df = pd.read_csv(target_csv)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst five rows:")
print(df.head())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
print("Available columns:", df.columns.tolist())
```

```python
def find_column(columns, candidates):
    for col in columns:
        if col.strip().lower() in candidates:
            return col
    return None

text_col = find_column(
    df.columns,
    {"text", "sentence", "text_sample", "content"}
)

language_col = find_column(
    df.columns,
    {"language", "lang", "label"}
)

if text_col is None or language_col is None:
    raise ValueError(
        "Could not identify the text and language columns. "
        "Check the printed column names and update text_col "
        "and language_col manually."
    )

df = df[[text_col, language_col]].copy()
df.columns = ["text", "language"]

# Handle missing values
df = df.dropna(subset=["text", "language"])

# Convert values to strings and remove extra whitespace
df["text"] = df["text"].astype(str).str.strip()
df["language"] = df["language"].astype(str).str.strip()

# Remove empty samples and duplicate text samples
df = df[
    (df["text"] != "") &
    (df["language"] != "")
]

df = df.drop_duplicates(subset=["text"]).reset_index(drop=True)

print("Cleaned dataset shape:", df.shape)
print("\nNumber of languages:", df["language"].nunique())
print("\nLanguage distribution:")
print(df["language"].value_counts())
```

```python
MAX_SAMPLES_PER_LANGUAGE = 500

df = (
    df.groupby("language", group_keys=False)
      .sample(
          n=MAX_SAMPLES_PER_LANGUAGE,
          random_state=42,
          replace=False
      )
      .reset_index(drop=True)
)

print("Subset shape:", df.shape)
print(df["language"].value_counts())
```

```python
df = (
    df.groupby("language", group_keys=False)
      .apply(
          lambda group: group.sample(
              n=min(len(group), 500),
              random_state=42
          )
      )
      .reset_index(drop=True)
)

print("Subset shape:", df.shape)
```

```python
df = (
    df.groupby("language", group_keys=False)
      .apply(
          lambda group: group.sample(
              n=min(len(group), 500),
              random_state=42
          )
      )
      .reset_index(drop=True)
)

print("Subset shape:", df.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    df["text"].values,
    df["language"].values,
    test_size=0.20,
    random_state=42,
    stratify=df["language"].values
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
MAX_WORDS = 20000
MAX_LEN = 100

tokenizer = Tokenizer(
    num_words=MAX_WORDS,
    oov_token="<OOV>",
    filters=""
)

tokenizer.fit_on_texts(X_train)

# Convert text into numerical sequences
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq = tokenizer.texts_to_sequences(X_test)

# Pad sequences to a fixed length
X_train_pad = pad_sequences(
    X_train_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

X_test_pad = pad_sequences(
    X_test_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

vocab_size = min(
    MAX_WORDS,
    len(tokenizer.word_index) + 1
)

print("Vocabulary size:", vocab_size)
print("Training sequence shape:", X_train_pad.shape)
print("Testing sequence shape:", X_test_pad.shape)
```

```python
# Calculate the number of unique classes (languages)
num_classes = len(np.unique(y_train))
print("Number of classes:", num_classes)

model = Sequential([
    Embedding(
        input_dim=vocab_size,
        output_dim=64,
        mask_zero=True
    ),

    LSTM(64),

    Dropout(0.3),

    Dense(32, activation="relu"),

    Dense(num_classes, activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
# Encode string labels into numerical integers
label_encoder = LabelEncoder()
y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

print("Encoded classes examples:", y_train_encoded[:5])
print("Class mapping:", dict(zip(label_encoder.classes_, range(len(label_encoder.classes_)))))

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True
)

history = model.fit(
    X_train_pad,
    y_train_encoded,
    epochs=10,
    batch_size=64,
    validation_split=0.1,
    callbacks=[early_stopping],
    verbose=1
)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("LSTM Model Accuracy")

plt.legend()
plt.grid(True)
plt.show()
```

```python
test_loss, test_accuracy = model.evaluate(
    X_test_pad,
    y_test_encoded,
    verbose=0
)

print("Test Loss:", round(test_loss, 4))
print("Test Accuracy:", round(test_accuracy, 4))
print("Test Accuracy (%):", round(test_accuracy * 100, 2))
```

```python
# 1. Predecir las probabilidades para el conjunto de prueba
y_pred_probs = model.predict(X_test_pad)
# 2. Obtener la clase con mayor probabilidad para cada muestra
y_pred_encoded = np.argmax(y_pred_probs, axis=1)

# 3. Decodificar las etiquetas numéricas a sus nombres de idiomas originales
language_names = list(label_encoder.classes_)

# 4. Mostrar el reporte de clasificación
print("=== Reporte de Clasificación ===")
print(classification_report(y_test_encoded, y_pred_encoded, target_names=language_names))

# 5. Calcular y graficar la matriz de confusión
cm = confusion_matrix(y_test_encoded, y_pred_encoded)
plt.figure(figsize=(15, 12))
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=language_names,
    yticklabels=language_names
)
plt.xlabel('Predicción')
plt.ylabel('Clase Real')
plt.title('Matriz de Confusión - Clasificador de Idiomas LSTM')
plt.show()
```

```python
y_prob = model.predict(X_test_pad, verbose=0)

y_pred = np.argmax(y_prob, axis=1)

# Usamos y_test_encoded en lugar de y_test para evitar mezclar strings e ints
accuracy = accuracy_score(y_test_encoded, y_pred)

precision = precision_score(
    y_test_encoded,
    y_pred,
    average="macro",
    zero_division=0
)

recall = recall_score(
    y_test_encoded,
    y_pred,
    average="macro",
    zero_division=0
)

f1 = f1_score(
    y_test_encoded,
    y_pred,
    average="macro",
    zero_division=0
)

print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1-Score :", round(f1, 4))
```

```python
print(
    classification_report(
        y_test_encoded,  # Cambiado de y_test a y_test_encoded
        y_pred,
        labels=np.arange(num_classes),
        target_names=label_encoder.classes_,
        zero_division=0
    )
)
```

```python
cm = confusion_matrix(
    y_test_encoded,  # Cambiado de y_test a y_test_encoded
    y_pred,
    labels=np.arange(num_classes)
)

plt.figure(figsize=(14, 12))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=label_encoder.classes_,
    yticklabels=label_encoder.classes_
)

plt.xlabel("Predicted Language")
plt.ylabel("Actual Language")
plt.title("Confusion Matrix - LSTM Language Identification")

plt.xticks(rotation=90)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
unseen_texts = [
    "The weather is beautiful today.",
    "I am learning machine learning.",
    "Please help me understand this problem.",
    "The students are studying in the library.",
    "Technology is changing the world."
]

# Convert unseen text into numerical sequences
unseen_seq = tokenizer.texts_to_sequences(unseen_texts)

unseen_pad = pad_sequences(
    unseen_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

# Predict language probabilities
unseen_prob = model.predict(unseen_pad, verbose=0)

predicted_labels = np.argmax(unseen_prob, axis=1)

predicted_languages = label_encoder.inverse_transform(
    predicted_labels
)

confidence_scores = np.max(unseen_prob, axis=1)

# Display predictions
for i, text_sample in enumerate(unseen_texts):
    print("Text:", text_sample)
    print("Predicted Language:", predicted_languages[i])
    print(
        "Model Confidence:",
        round(float(confidence_scores[i]) * 100, 2),
        "%"
    )
    print("-" * 50)
```

```python
# Predict the language of five unseen text samples

unseen_texts = [
    "The weather is beautiful today.",
    "I am learning machine learning.",
    "Please help me understand this problem.",
    "The students are studying in the library.",
    "Technology is changing the world."
]

# Convert text into sequences
unseen_seq = tokenizer.texts_to_sequences(unseen_texts)

# Pad sequences
unseen_pad = pad_sequences(
    unseen_seq,
    maxlen=MAX_LEN,
    padding="post",
    truncating="post"
)

# Predict language probabilities
predictions = model.predict(unseen_pad, verbose=0)

# Display results
for text, probabilities in zip(unseen_texts, predictions):
    predicted_index = np.argmax(probabilities)
    predicted_language = label_encoder.inverse_transform(
        [predicted_index]
    )[0]

    confidence = probabilities[predicted_index] * 100

    print("Text:", text)
    print("Predicted Language:", predicted_language)
    print("Model Confidence: {:.2f}%".format(confidence))
    print("-" * 50)
```
